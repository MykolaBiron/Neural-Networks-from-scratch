## Build the simplest neural network with 2 fully connected layers

```python
import numpy as np
import sys
sys.path.append("..")
from utils.DenseLayer import DenseLayer
from utils.Tanh import Tanh
from utils.Relu import Relu
from utils.activation_funcs import mse, mse_prime
```

```python
# Define data X - inputs, Y - outputs
X = np.reshape([[0, 0], [0, 1], [1, 0], [1, 1]], (4, 2, 1))
Y = np.reshape([[0], [1], [1], [0]], (4, 1, 1))


# Create a network by stacking the layers and applying the activation function after each
dense_nn = [
    DenseLayer(2, 5),
    Tanh(),
    DenseLayer(5, 1),
    Tanh(),
]

epochs = 10000 # The number of iterations / weight updates
learning_rate = 0.1 # Learning rate controls how much weights are adjusted during backpropagation

for epoch in range(epochs):
    err = 0
    for (x, y) in zip(X, Y):
        output = x
        for layer in dense_nn:
            output = layer.forward(output)
        err += mse(y, output)

        grad = mse_prime(y, output)
        for layer in reversed(dense_nn):
            grad = layer.backward(grad, learning_rate)

    err /= len(X)
    
    # Print every 1000 epochs to avoid clutter
    if epoch % 1000 == 0:
        print(f"Epoch {epoch:5d} | Error: {err[0][0]:.6f}")

print(f"\nFinal Error: {err[0][0]:.6f}")

# Test the network
print("\nTesting XOR predictions:")
for x, y in zip(X, Y):
    output = x
    for layer in dense_nn:
        output = layer.forward(output)
    print(f"Input: {x.T[0]} | Expected: {y[0][0]} | Predicted: {output[0][0]:.4f}")
```
